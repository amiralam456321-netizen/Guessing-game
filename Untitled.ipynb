**Guessing game**

```python
import random

print("=== Guessing Game with Levels ===")
print("1. Easy   (1-50, 5 attempts)")
print("2. Medium (1-100, 7 attempts)")
print("3. Hard   (1-200, 10 attempts)")

levels = {
    "1": (50, 5),
    "2": (100, 7),
    "3": (200, 10),
}

choice = input("Choose a level (1/2/3): ").strip()
while choice not in levels:
    choice = input("Invalid choice. Choose 1, 2 or 3: ").strip()

max_number, max_attempts = levels[choice]
number = random.randint(1, max_number)
attempts = 0

print(f"\nI'm thinking of a number between 1 and {max_number}.")
print(f"You have {max_attempts} attempts.\n")

while attempts < max_attempts:
    guess = input(f"Attempt {attempts + 1}/{max_attempts} - your guess: ").strip()

    if not guess.isdigit():
        print("Please enter a valid number!\n")
        continue

    guess = int(guess)
    attempts += 1

    if guess > number:
        print("Too low!\n")
    elif guess < number:
        print("Too high!\n")
    else:
        print(f"Correct! You guessed {number} in {attempts} attempts. You win!")
        break
else:
    print(f"Out of attempts! The number was {number}.")
```
