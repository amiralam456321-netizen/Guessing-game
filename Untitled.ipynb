**Guessing game**

```python
print("=== I Will Guess Your Number ===")
print("Think of a number between 1 and 100. Don't tell me!")
input("Press Enter when you're ready...\n")

low = 1
high = 100
attempts = 0

while low <= high:
    guess = (low + high) // 2
    attempts += 1

    answer = input(f"Is your number {guess}? (h = too high, l = too low, c = correct): ").lower().strip()

    if answer == "c":
        print(f"\nYay! I guessed your number {guess} in {attempts} attempts!")
        break
    elif answer == "h":
        high = guess - 1
    elif answer == "l":
        low = guess + 1
    else:
        print("Please type h, l or c.\n")
        attempts -= 1
else:
    print("\nHmm, your answers don't add up. Did you change your number?")
```
